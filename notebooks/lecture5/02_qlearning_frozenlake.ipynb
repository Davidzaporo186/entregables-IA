## 1. Imports y funciones de Q-Learning


```python
import numpy as np
import gymnasium as gym
import random
import matplotlib.pyplot as plt
import time
from IPython.display import clear_output, display

from IPython.display import HTML
from matplotlib import animation
import matplotlib.pyplot as plt

def play_episode(env, Q=None, random_policy=False, max_steps=100, seed=None):
    """Ejecuta un episodio y devuelve sus frames y recompensa total."""
    state, _ = env.reset(seed=seed)
    frames = [env.render()]
    total_reward = 0.0

    for _ in range(max_steps):
        if random_policy:
            action = env.action_space.sample()
        else:
            q_values = Q[state]
            max_q = np.max(q_values)

            best_actions = np.flatnonzero(q_values == max_q)
            action = int(np.random.choice(best_actions))

        next_state, reward, terminated, truncated, _ = env.step(action)

        frames.append(env.render())
        total_reward += reward
        state = next_state

        if terminated or truncated:
            break

    return frames, total_reward

def frames_to_video(frames, interval=700):
    """Convierte una lista de frames RGB en una animación reproducible en Jupyter."""
    fig = plt.figure(figsize=(4, 4))
    plt.axis("off")

    image = plt.imshow(frames[0])

    def update(frame):
        image.set_data(frame)
        return [image]

    anim = animation.FuncAnimation(
        fig,
        update,
        frames=frames,
        interval=interval,
        blit=True,
        repeat=True
    )

    plt.close(fig)
    return HTML(anim.to_jshtml())

def initialize_q_table(state_space, action_space):
    return np.zeros((state_space, action_space))

def greedy_policy(Qtable, state):
    max_q = np.max(Qtable[state])
    best_actions = np.flatnonzero(Qtable[state] == max_q)
    return int(np.random.choice(best_actions))

def epsilon_greedy_policy(Qtable, state, epsilon, env):
    if random.random() < epsilon:
        return env.action_space.sample()

    return greedy_policy(Qtable, state)

def train_q_learning(
    env,
    Qtable,
    n_episodes=5000,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.001,
    max_steps=100,
    start_episode=0,
):
    rewards = []

    for episode in range(n_episodes):
        state, _ = env.reset()
        total_reward = 0

        global_episode = start_episode + episode
        epsilon = min_epsilon + (
            max_epsilon - min_epsilon
        ) * np.exp(-decay_rate * global_episode)

        for _ in range(max_steps):
            action = epsilon_greedy_policy(
                Qtable, state, epsilon, env
            )

            next_state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated

            best_next_q = 0.0 if done else np.max(Qtable[next_state])

            td_target = reward + gamma * best_next_q
            td_error = td_target - Qtable[state, action]

            Qtable[state, action] += learning_rate * td_error

            state = next_state
            total_reward += reward

            if done:
                break

        rewards.append(total_reward)

    return Qtable, rewards

def evaluate_q_policy(env, Qtable, n_episodes=100, max_steps=100):
    episode_rewards = []

    for _ in range(n_episodes):
        state, _ = env.reset()
        total_reward = 0

        for _ in range(max_steps):
            action = greedy_policy(Qtable, state)

            next_state, reward, terminated, truncated, _ = env.step(action)

            state = next_state
            total_reward += reward

            if terminated or truncated:
                break

        episode_rewards.append(total_reward)

    return np.mean(episode_rewards), np.std(episode_rewards)

def show_frame(env, title=''):
    frame = env.render()
    plt.figure(figsize=(4, 4))
    plt.imshow(frame)
    plt.axis('off')
    plt.title(title)
    display(plt.gcf())
    plt.close()
```

### 💬 Antes de ejecutar

En CheeseWorld teníamos explícitamente una clase `Environment` y una clase `QLearningAgent`.

**Pregunta:** en este notebook, ¿qué papel cumple Gymnasium y dónde quedó representado el agente?

**Notas:**

- Gymnasium hace el papel del `Environment` de CheeseWorld: define los estados, las acciones, la dinámica y la recompensa. Nosotros solo llamamos a `reset()`, `step()` y `render()`.
- El agente no quedó en una clase aparte. Está repartido en lo que escribimos acá: la Q-table, `epsilon_greedy_policy` (cómo decide) y `train_q_learning` (cómo actualiza con el error TD).
- Gymnasium no sabe nada de Q-Learning; lo único que hace es devolvernos $(s', r, done)$ después de cada acción.


## 2. Crear FrozenLake


```python
env = gym.make(
    "FrozenLake-v1",
    map_name="4x4",
    is_slippery=False,
    render_mode="rgb_array"
)

random.seed(42)
np.random.seed(42)
env.action_space.seed(42)
state, info = env.reset(seed=42)

print("Estado inicial:", state)
print("Número de estados:", env.observation_space.n)
print("Número de acciones:", env.action_space.n)
```

En FrozenLake las acciones son:

| Acción | Código |
|---|---:|
| Left | 0 |
| Down | 1 |
| Right | 2 |
| Up | 3 |

Primero trabajaremos con `is_slippery=False`, es decir, con transiciones determinísticas.


### 💬 Actividad 1 — La Q-table

Antes de crearla:

1. ¿Cuántas filas debe tener la Q-table?
2. ¿Cuántas columnas?
3. ¿Qué representa una celda \(Q[s,a]\)?

**Respuesta / discusión:**

- Filas: **16**, una por estado (el tablero es de 4×4).
- Columnas: **4**, una por acción (Left, Down, Right, Up).
- $Q[s,a]$ es la recompensa futura descontada que el agente espera obtener si en el estado $s$ toma la acción $a$ y después sigue actuando lo mejor posible.


```python
state_space = env.observation_space.n
action_space = env.action_space.n

Q = initialize_q_table(state_space, action_space)

print("Q-table shape:", Q.shape)
Q
```

### 💬 Actividad 2 — Inicio del aprendizaje

Todos los valores son cero.

$$
Q(s,a)=0
$$

¿Esto significa que todas las acciones son malas, o que el agente todavía no sabe nada?

¿Qué ocurre si varias acciones tienen exactamente el mismo valor máximo?

**Notas:**

- Que todo sea cero no quiere decir que las acciones sean malas, sino que el agente todavía no sabe nada: no ha recibido ninguna experiencia que le diga qué tan buena es cada una.
- Si hay varias acciones empatadas en el máximo, `greedy_policy` elige una al azar entre las empatadas. Por eso al inicio el agente se comporta como una caminata aleatoria. Si usáramos un `argmax` normal, siempre escogería la primera (Left) y se quedaría pegado contra la pared.


## 3. Ejecutar una transición


```python
state, _ = env.reset(seed=7)

epsilon = 1.0
action = epsilon_greedy_policy(Q, state, epsilon, env)

next_state, reward, terminated, truncated, _ = env.step(action)

print("state      =", state)
print("action     =", action)
print("reward     =", reward)
print("next_state =", next_state)
print("done       =", terminated or truncated)
```

### 💬 Actividad 3 — Identificar la experiencia

Escribe la experiencia anterior como:

$$
(s,a,r,s')
$$

**Experiencia:**

$$
(0,\ 0,\ 0,\ 0)
$$

(estado 0, acción Left, recompensa 0, siguiente estado 0: el agente intentó ir a la izquierda desde la esquina y chocó con el borde, así que no se movió).

En Value Iteration no teníamos como tal el **siguiente estado $s'$ observado**. Allí conocíamos el modelo $T(s,a,s')$ completo y hacíamos la suma sobre todos los $s'$ posibles con sus probabilidades. Con experiencia real solo vemos **un** $s'$ por intento (y la recompensa $r$ nos la da el ambiente al interactuar, no una tabla $R(s)$ que ya conocemos).

¿De cuál de esos cuatro elementos **no disponíamos directamente** en Value Iteration cuando hablábamos de experiencia real?


## 4. Del azar a una política aprendida

Vamos a observar **el mismo agente en tres momentos**. Primero no sabe nada y actúa al azar; luego veremos su política después de pocas experiencias; finalmente veremos la política después del entrenamiento completo.


```python
Q_initial = initialize_q_table(
    env.observation_space.n,
    env.action_space.n
)

EARLY_EPISODES = 50
Q_early = Q_initial.copy()
Q_early, rewards_early = train_q_learning(
    env,
    Q_early,
    n_episodes=EARLY_EPISODES,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.001,
)

TOTAL_EPISODES = 10000
Q_trained = Q_early.copy()
Q_trained, rewards_final = train_q_learning(
    env,
    Q_trained,
    n_episodes=TOTAL_EPISODES - EARLY_EPISODES,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.001,
    start_episode=EARLY_EPISODES,
)

Q = Q_trained
rewards = rewards_early + rewards_final

print('Snapshots guardados:')
print('Q_initial : 0 episodios')
print(f'Q_early   : {EARLY_EPISODES} episodios')
print(f'Q_trained : {TOTAL_EPISODES} episodios')
```

### Momento 1 — Sin entrenamiento: random walk
Todavía no usamos la Q-table para decidir. Cada acción se selecciona aleatoriamente. Observa cómo interactúa el agente con el mundo.

```python
frames_random, reward_random = play_episode(
    env,
    Q_initial,
    random_policy=True,
    seed=7
)

print(f"Recompensa total: {reward_random}")
frames_to_video(frames_random, interval=700)
```

### Momento 2 — Después de pocas iteraciones

Ahora el agente usa de forma **greedy** lo que ha aprendido en `Q_early`. Todavía conoce poco del ambiente, así que su comportamiento puede ser incompleto o equivocarse.


```python
frames_early, reward_early = play_episode(
    env,
    Q_early,
    random_policy=False,
    seed=7
)

print(f"Recompensa total: {reward_early}")
frames_to_video(frames_early, interval=700)
```

### Momento 3 — Agente entrenado

Finalmente usamos `Q_trained`. Ya no exploramos: en cada estado el agente selecciona una de las acciones con mayor valor $Q(s,a)$.


```python
frames_trained, reward_trained = play_episode(
    env,
    Q_trained,
    random_policy=False,
    seed=7
)

print(f"Recompensa total: {reward_trained}")
frames_to_video(frames_trained, interval=700)
```

### 💬 Actividad — ¿Qué cambió?

Compara las tres ejecuciones. El ambiente, los estados y las acciones son los mismos. **¿Qué cambió internamente en el agente para que su comportamiento mejore?**

Observa `Q_initial`, `Q_early` y `Q_trained` y relaciona sus valores con las acciones que viste ejecutar.


**Respuesta:**

Lo único que cambió dentro del agente es la **Q-table**; el ambiente, los estados y las acciones son los mismos.

- `Q_initial` es todo ceros: no hay preferencia entre acciones y el agente camina al azar (recompensa 0 en el episodio de prueba).
- `Q_early` (50 episodios) casi no sabe nada: en mi ejecución solo el estado 14 tiene valores distintos de cero (`Q(14, Right) = 0.7`, el paso que llega a la meta), y los otros 15 estados siguen en cero. Ojo: aunque ese episodio llegó a la meta, fue en buena parte suerte, porque en los estados con Q en cero el desempate sigue siendo aleatorio.
- `Q_trained` (10 000 episodios) tiene valores que "se propagaron hacia atrás" desde la meta: cada paso más lejos vale un poco menos por el descuento $\gamma = 0.95$ (por ejemplo, $Q(0,\text{Down}) \approx 0.95^5 \approx 0.774$ porque la meta está a 6 pasos). Siguiendo la acción de mayor Q en cada estado, el agente va directo a la meta.

### 💬 Actividad 4 — Leer una fila de Q

Selecciona un estado $s$ y observa:

$$
Q(s,0), Q(s,1), Q(s,2), Q(s,3)
$$

**Estado seleccionado:** 14 (la celda justo a la izquierda de la meta)

**Valores Q:**

- Left: 0.902
- Down: 0.950
- Right: 1.000
- Up: 0.902

¿Cuál acción seleccionaría:

$$
\arg\max_a Q(s,a)
$$

?

**Interpretación:**

> $\arg\max_a Q(14,a) = $ **Right** (Q = 1.0): desde ahí un solo paso lleva a la meta y da recompensa 1. *Down* vale 0.95 porque en la fila de abajo no hay más tablero, así que el agente se queda en el mismo estado y pierde un paso (1.0 descontado una vez por $\gamma$). *Left* y *Up* valen 0.902 porque se alejan de la meta y hay que volver a pasar por el estado 14, lo que cuesta dos pasos de descuento ($0.95^2$).


```python
estado = 14
nombres = ["Left", "Down", "Right", "Up"]
for nombre, a in zip(nombres, range(4)):
    print(f"Q({estado},{a}) {nombre:5s}: initial={Q_initial[estado,a]:.3f} | early={Q_early[estado,a]:.3f} | trained={Q_trained[estado,a]:.3f}")
print("argmax trained:", nombres[int(np.argmax(Q_trained[estado]))])

print()
for nombre, a in zip(nombres, range(4)):
    print(f"Q(0,{a}) {nombre:5s}: trained={Q_trained[0,a]:.3f}")
print("Estados con Q todo en cero en Q_trained:", int((Q_trained.max(axis=1)==0).sum()))
print("Estados con Q todo en cero en Q_early  :", int((Q_early.max(axis=1)==0).sum()))
```

## 5. Evaluar la política aprendida


```python
mean_reward, std_reward = evaluate_q_policy(
    env,
    Q_trained,
    n_episodes=100
)

print(f"Mean reward: {mean_reward:.3f}")
print(f"Std reward : {std_reward:.3f}")
```

### 💬 Actividad 5 — Exploration vs. exploitation

Durante entrenamiento usamos $\epsilon$-greedy.

Durante evaluación usamos:

$$
a=\arg\max_aQ(s,a)
$$

¿Por qué **no exploramos** durante la evaluación?

**Conclusión:**

> En la evaluación queremos medir qué tan buena es la política que ya aprendió, no seguir aprendiendo. Si exploráramos, una parte de las acciones sería aleatoria y la recompensa bajaría por culpa de esa aleatoriedad y no de la calidad de la Q-table. La exploración sirve para descubrir información mientras se entrena; al evaluar, explotamos lo aprendido. Con `is_slippery=False` la política entrenada llega a la meta siempre (mean reward = 1.0, std = 0.0).


## 6. Experimento: FrozenLake estocástico


```python
slippery_env = gym.make(
    "FrozenLake-v1",
    map_name="4x4",
    is_slippery=True,
    render_mode="rgb_array"
)

slippery_env.reset(seed=42)
slippery_env.action_space.seed(42)

Q_slippery = initialize_q_table(
    slippery_env.observation_space.n,
    slippery_env.action_space.n
)

Q_slippery, rewards_slippery = train_q_learning(
    slippery_env,
    Q_slippery,
    n_episodes=20000,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.0005,
    max_steps=100
)

mean_reward, std_reward = evaluate_q_policy(
    slippery_env,
    Q_slippery,
    n_episodes=500
)

print(f"Mean reward: {mean_reward:.3f}")
print(f"Std reward : {std_reward:.3f}")
```

### 💬 Actividad 6 — Determinístico vs. estocástico

Compara:

- `is_slippery=False`
- `is_slippery=True`

¿Qué cambia en el **ambiente**?

¿Qué cambia en la **ecuación de Q-Learning**?

**Discusión:**

- Ambiente: con `is_slippery=True` el ambiente deja de ser determinístico. La acción elegida solo se cumple una parte de las veces y en las demás el agente se desliza en una dirección perpendicular, así que la misma pareja $(s,a)$ puede terminar en distintos $s'$, es decir, $P(s'\mid s,a)$ ya no es 0 o 1.
- Algoritmo: la ecuación de actualización **es la misma**, $Q \leftarrow Q + \alpha\,[r+\gamma\max_{a'}Q(s',a')-Q]$. Lo que cambia es la experiencia que entra: como el mismo $(s,a)$ da resultados distintos, $Q(s,a)$ pasa a aproximar un **promedio** sobre las posibles transiciones y se necesitan más episodios (usamos 20 000 y un decaimiento de $\epsilon$ más lento) para que se asiente. Aun así el desempeño tiene un techo: en la evaluación el agente logró un mean reward de 0.702 (std 0.457), o sea que llega a la meta en cerca del 70 % de los episodios y no en todos, a diferencia del caso determinístico (1.0).


# Cierre de clase

Completa antes de terminar:

**1. ¿Qué almacena $Q(s,a)$?**

> La recompensa futura acumulada y descontada que se espera obtener al tomar la acción $a$ en el estado $s$ y luego seguir la mejor política.

**2. ¿De dónde sale $\max_{a'}Q(s',a')$?**

> De la misma Q-table: se mira la fila del siguiente estado $s'$ y se toma el mayor valor. Es una estimación actual (por eso se dice que hace *bootstrapping*); en un estado terminal vale 0.

**3. ¿Por qué necesitamos $\epsilon$-greedy?**

> Para balancear exploración y explotación. Si el agente solo explotara, se quedaría con la primera ruta que encuentre (o con ceros, sin saber que hay algo mejor); con algo de acciones aleatorias descubre caminos nuevos y corrige sus valores.

**4. ¿Por qué Q-Learning es model-free?**

> Porque nunca usa $T(s,a,s')$ ni la función de recompensa; solo aprende de las transiciones $(s,a,r,s')$ que observa al interactuar con el ambiente.
